```python
coverage_df['TARGET_NUMBER'] = coverage_df['TARGET_NUMBER'].fillna(coverage_df['TARGET_NUMBER'].median())
coverage_df['DOSES'] = coverage_df['DOSES'].fillna(coverage_df['DOSES'].median())
coverage_df['COVERAGE'] = coverage_df['COVERAGE'].fillna(coverage_df['COVERAGE'].median())

coverage_df['NAME'] = coverage_df['NAME'].fillna(
    coverage_df['NAME'].mode()[0]
)
```

```python
coverage_df.isnull().sum()
```

```python
incidence_df['INCIDENCE_RATE'] = incidence_df['INCIDENCE_RATE'].fillna(0)
```

```python
incidence_df.isnull().sum()
```

```python
reported_cases_df['CASES'] = reported_cases_df['CASES'].fillna(0)
reported_cases_df.isnull().sum()
```

```python
vaccine_intro_df[['COUNTRYNAME','WHO_REGION','DESCRIPTION','INTRO']] = (
    vaccine_intro_df[['COUNTRYNAME','WHO_REGION','DESCRIPTION','INTRO']]
    .fillna('Unknown')
)
vaccine_intro_df.isnull().sum()
```

```python
vaccine_schedule_df[['TARGETPOP','AGEADMINISTERED','SOURCECOMMENT','GEOAREA']] = (
    vaccine_schedule_df[['TARGETPOP','AGEADMINISTERED','SOURCECOMMENT','GEOAREA']]
    .fillna('Not Specified')
)
vaccine_schedule_df.isnull().sum()
```

```python
for df in [coverage_df, incidence_df, reported_cases_df, vaccine_intro_df, vaccine_schedule_df]:
    df.drop_duplicates(inplace=True)
```

```python
coverage_df.dropna(subset=['YEAR', 'CODE'], inplace=True)
coverage_df.isnull().sum()
```

```python
incidence_df.dropna(subset=['YEAR', 'CODE'], inplace=True)
incidence_df.isnull().sum()
```

```python
reported_cases_df.dropna(subset=['YEAR', 'CODE'], inplace=True)
reported_cases_df.isnull().sum()
```

```python
vaccine_intro_df.dropna(subset=['YEAR'], inplace=True)
vaccine_intro_df.isnull().sum()
```

```python
vaccine_schedule_df.dropna(
    subset=['YEAR', 'VACCINECODE', 'COUNTRYNAME'],
    inplace=True
)
vaccine_schedule_df.isnull().sum()
```

## Has vaccination coverage improved over time?

```python
coverage_trend = coverage_df.groupby('YEAR')['COVERAGE'].mean()

plt.figure(figsize=(10,5))
plt.plot(coverage_trend.index, coverage_trend.values)
plt.xlabel("Year")
plt.ylabel("Average Vaccination Coverage (%)")
plt.title("Global Vaccination Coverage Trend")
plt.show()
```

## Are disease cases decreasing?

```python
incidence_trend = incidence_df.groupby('YEAR')['INCIDENCE_RATE'].mean()

plt.figure(figsize=(10,5))
plt.plot(incidence_trend.index, incidence_trend.values)
plt.xlabel("Year")
plt.ylabel("Average Incidence Rate")
plt.title("Global Disease Incidence Trend")
plt.show()
```

## Do higher vaccination rates reduce disease incidence?

```python
merged_df = pd.merge(
    coverage_df,
    incidence_df,
    on=['CODE','YEAR'],
    how='inner'
)

plt.figure(figsize=(7,5))
sns.scatterplot(
    data=merged_df,
    x='COVERAGE',
    y='INCIDENCE_RATE'
)
plt.xlabel("Vaccination Coverage (%)")
plt.ylabel("Disease Incidence Rate")
plt.title("Vaccination Coverage vs Disease Incidence")
plt.show()
```

## Which diseases are most prevalent?

```python
top_diseases = reported_cases_df.groupby('DISEASE_DESCRIPTION')['CASES'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(8,5))
top_diseases.plot(kind='bar')
plt.xlabel("Disease")
plt.ylabel("Total Reported Cases")
plt.title("Top 10 Diseases by Reported Cases")
plt.show()
```

## Regional Coverage Comparison

```python
region_coverage = coverage_df.groupby('NAME')['COVERAGE'].mean().sort_values(ascending=False).head(10)
region_coverage


```

## Which regions / countries have higher or lower vaccination coverage?

```python
# Average coverage by country
country_coverage = (
    coverage_df
    .groupby('NAME')['COVERAGE']
    .mean()
    .sort_values(ascending=False)
)

# Top 10 countries
top_10 = country_coverage.head(10)

plt.figure(figsize=(10,5))
top_10.plot(kind='bar')
plt.xlabel("Country")
plt.ylabel("Average Vaccination Coverage (%)")
plt.title("Top 10 Countries by Vaccination Coverage")
plt.xticks(rotation=45)
plt.show()
```

## Bottom 10 Countries

### “Identify regions with low vaccination coverage”

```python
bottom_10 = country_coverage.tail(10)

plt.figure(figsize=(10,5))
bottom_10.plot(kind='bar')
plt.xlabel("Country")
plt.ylabel("Average Vaccination Coverage (%)")
plt.title("Bottom 10 Countries by Vaccination Coverage")
plt.xticks(rotation=45)
plt.show()
```

#### “The bar charts highlight countries with the highest and lowest average vaccination coverage, helping identify regions that may require targeted public health interventions.”

## Country vs Year Heatmap (Coverage Trend)

```python
pivot_coverage = coverage_df.pivot_table(
    values='COVERAGE',
    index='NAME',
    columns='YEAR',
    aggfunc='mean'
)

plt.figure(figsize=(12,6))
sns.heatmap(pivot_coverage.head(15), cmap='viridis')
plt.title("Vaccination Coverage Heatmap (Sample Countries)")
plt.xlabel("Year")
plt.ylabel("Country")
plt.show()
```
