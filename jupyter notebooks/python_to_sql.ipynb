```python
engine = create_engine(
    "mysql+mysqlconnector://root:Pari1503@localhost:3306/vaccination_analysis"
)
```

```python
coverage_df = coverage_df.rename(columns={
    'GROUP': 'group_name',
    'CODE': 'country_code',
    'NAME': 'country_name',
    'YEAR': 'year',
    'ANTIGEN': 'antigen',
    'ANTIGEN_DESCRIPTION': 'antigen_description',
    'COVERAGE_CATEGORY': 'coverage_category',
    'COVERAGE_CATEGORY_DESCRIPTION': 'coverage_category_description',
    'TARGET_NUMBER': 'target_number',
    'DOSES': 'doses',
    'COVERAGE': 'coverage'
})
```

```python
incidence_df = incidence_df.rename(columns={
    'GROUP': 'group_name',
    'CODE': 'country_code',
    'NAME': 'country_name',
    'YEAR': 'year',
    'DISEASE': 'disease',
    'DISEASE_DESCRIPTION': 'disease_description',
    'DENOMINATOR': 'denominator',
    'INCIDENCE_RATE': 'incidence_rate'
})
```

```python
reported_cases_df = reported_cases_df.rename(columns={
    'GROUP': 'group_name',
    'CODE': 'country_code',
    'NAME': 'country_name',
    'YEAR': 'year',
    'DISEASE': 'disease',
    'DISEASE_DESCRIPTION': 'disease_description',
    'CASES': 'cases'
})
```

```python
vaccine_intro_df = vaccine_intro_df.rename(columns={
    'ISO_3_CODE': 'iso_3_code',
    'COUNTRYNAME': 'country_name',
    'WHO_REGION': 'who_region',
    'YEAR': 'year',
    'DESCRIPTION': 'description',
    'INTRO': 'intro'
})
```

```python
vaccine_schedule_df = vaccine_schedule_df.rename(columns={
    'ISO_3_CODE': 'iso_3_code',
    'COUNTRYNAME': 'country_name',
    'WHO_REGION': 'who_region',
    'YEAR': 'year',
    'VACCINECODE': 'vaccine_code',
    'VACCINE_DESCRIPTION': 'vaccine_description',
    'SCHEDULEROUNDS': 'schedule_rounds',
    'TARGETPOP': 'target_pop',
    'TARGETPOP_DESCRIPTION': 'target_pop_description',
    'GEOAREA': 'geoarea',
    'AGEADMINISTERED': 'age_administered',
    'SOURCECOMMENT': 'source_comment'
})
```

```python
# Find all rows where country_code is longer than 20 characters
long_codes = coverage_df[coverage_df['country_code'].str.len() > 20]

print("Rows with long country_code values:")
print(long_codes[['country_code']])
coverage_df['country_code'] = coverage_df['country_code'].str[:20]

```

```python
coverage_df.to_sql(
    'coverage_data',
    engine,
    if_exists='append',
    index=False,
    chunksize=10000
)
```

```python
incidence_df.to_sql(
    name='incidence_rate',
    con=engine,
    if_exists='append',
    index=False
)
```
